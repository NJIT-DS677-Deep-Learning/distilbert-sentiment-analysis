# TweetEval Sentiment Dataset Test

Run the cells in order to download the [TweetEval dataset](https://huggingface.co/datasets/cardiffnlp/tweet_eval) using its `sentiment` configuration and print two training examples. The dataset is cached locally for subsequent runs.

Sentiment labels are negative, neutral, and positive; the label names below come from the dataset metadata.

## Load the tweet_eval dataset with the sentiment config

```python
from datasets import load_dataset

dataset = load_dataset("cardiffnlp/tweet_eval", "sentiment")
print(dataset)
```

### Print some sample values

Possible values are 0 - Negative, 1 - Neutral, 2 - Positive

This dataset does not mention if it has sarcasm or mixed views, it is useful as an additional sentiment analysis dataset

```python
train_data = dataset["train"]
label_names = train_data.features["label"].names

for row in train_data.select(range(20)):
    print(f"Text: {row['text']}")
    print(f"Label: {row['label']} ({label_names[row['label']]})")
    print("-" * 70)
```
