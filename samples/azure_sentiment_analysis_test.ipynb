```python
import requests
import json
import os

endpoint = "https://njit-sentiment.cognitiveservices.azure.com/"
api_key = os.environ["NJIT_AZURE_LANGUAGE_KEY"]

url = f"{endpoint}/language/:analyze-text?api-version=2023-04-01"

payload = {
    "kind": "SentimentAnalysis",
    "analysisInput": {
        "documents": [
            {
                "id": "1",
                "language": "en",
                "text": "I absolutely love this product. It works great!"
            }
        ]
    }
}

headers = {
    "Ocp-Apim-Subscription-Key": api_key,
    "Content-Type": "application/json"
}

response = requests.post(url, headers=headers, json=payload)
print(json.dumps(response.json(), indent=4))
```
