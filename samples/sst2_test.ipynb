# SST-2 Dataset Sample

What do the text and sentiment labels in SST-2 look like? Run the cells in order to load the training split of [Stanford SST-2](https://huggingface.co/datasets/stanfordnlp/sst2) and display five examples.

SST-2 contains movie-review text with binary sentiment labels: `0` is negative and `1` is positive. There is no neutral class. The training split is used because the public test split hides its sentiment labels.

```python
from datasets import load_dataset

DATASET_ID = "stanfordnlp/sst2"
SPLIT = "train"
SAMPLE_SIZE = 5
SEED = 42

dataset = load_dataset(DATASET_ID, split=SPLIT)
print(f"Dataset: {DATASET_ID} | Split: {SPLIT} | Rows: {len(dataset):,}")
print(dataset.features)
```

## Preview five training examples

Shuffle with seed `42` to make the sample reproducible. Print each example's original index, full text, integer label, and sentiment name from the dataset metadata. This small preview is not a class-balance estimate.

```python
sample_data = dataset.shuffle(seed=SEED).select(range(SAMPLE_SIZE))
label_names = dataset.features["label"].names

for number, row in enumerate(sample_data, start=1):
    print(f"Example {number} | Index: {row['idx']}")
    print(f"Text: {row['sentence']}")
    print(f"Sentiment: {row['label']} ({label_names[row['label']]})")
    print("-" * 70)
```
